## Module 1 – NumPy Operations

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Mean:", np.mean(arr))
print("Sum:", np.sum(arr))
print("Multiplication:", arr * 2)
```

## Module 2 – Pandas Operations

```python
import pandas as pd

student_data = {
    'Name': ['Aman', 'Riya', 'Karan', 'Priya'],
    'Marks': [85, 90, 78, 92]
}

students = pd.DataFrame(student_data)

print(students)
print("Average Marks:", students['Marks'].mean())
```

## Module 3 – Matplotlib Visualization

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

**Expected output (from the Word document):** A line graph is displayed showing values of x and y.

## Module 4 – TensorFlow Tensor Operations

```python
import tensorflow as tf

tensor_tf = tf.constant([[1, 2], [3, 4]])

print(tensor_tf)
```

## Module 5 – PyTorch Tensor Operations

```python
import torch

tensor_torch = torch.tensor([[5, 6], [7, 8]])

print(tensor_torch)
```

## Module 1 – NumPy Operations

### Theory
NumPy (Numerical Python) is a powerful Python library used for numerical and mathematical operations. It provides support for multidimensional arrays, matrices, and various mathematical functions. NumPy is faster than traditional Python lists because operations are optimized internally. It is widely used in data science, machine learning, and scientific computing.

### Common Syntax
```python
import numpy as np
arr = np.array([1, 2, 3])
np.mean(arr)
np.sum(arr)
```

````python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Mean:", np.mean(arr))
print("Sum:", np.sum(arr))
print("Multiplication:", arr * 2)
````

## Module 2 – Pandas Operations

### Theory
Pandas is a Python library used for data manipulation and analysis. It provides two major data structures: Series and DataFrame. Pandas helps in handling tabular data efficiently, performing data cleaning, filtering, grouping, and statistical analysis. It is widely used in machine learning and data analytics projects.

### Common Syntax
```python
import pandas as pd
data = pd.DataFrame()
data.head()
data.mean()
```

````python
import pandas as pd

student_data = {
    'Name': ['Aman', 'Riya', 'Karan', 'Priya'],
    'Marks': [85, 90, 78, 92]
}

students = pd.DataFrame(student_data)

print(students)
print("Average Marks:", students['Marks'].mean())
````

## Module 3 – Matplotlib Visualization

### Theory
Matplotlib is a visualization library in Python used to create charts, graphs, and plots. It helps in representing data visually for better understanding and analysis. Different types of plots such as line plots, bar charts, histograms, and pie charts can be created using Matplotlib. It is widely used in data science and analytics.

### Common Syntax
```python
import matplotlib.pyplot as plt
plt.plot(x, y)
plt.title("Graph")
plt.show()
```

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
````

**Expected output (from the Word document):** A line graph is displayed showing values of x and y.

## Module 4 – TensorFlow Tensor Operations

### Theory
TensorFlow is an open-source deep learning framework developed by Google. It is mainly used for machine learning and artificial intelligence applications. TensorFlow uses tensors, which are multidimensional arrays used for computations. It supports neural network development, training, and deployment.

### Common Syntax
```python
import tensorflow as tf
tensor = tf.constant([[1,2],[3,4]])
print(tensor)
```

````python
import tensorflow as tf

tensor_tf = tf.constant([[1, 2], [3, 4]])

print(tensor_tf)
````

## Module 5 – PyTorch Tensor Operations

### Theory
PyTorch is an open-source machine learning library developed by Meta AI. It is widely used for deep learning and research applications. PyTorch provides dynamic computation graphs and efficient tensor operations. It is popular because of its flexibility, simplicity, and strong GPU support.

### Common Syntax
```python
import torch
tensor = torch.tensor([[1,2],[3,4]])
print(tensor)
```

````python
import torch

tensor_torch = torch.tensor([[5, 6], [7, 8]])

print(tensor_torch)
````

## Theory
A Perceptron is the simplest type of artificial neural network used for binary classification problems. It takes multiple inputs, applies weights, adds bias, and passes the result through an activation function to generate output.

The perceptron follows the equation:

$$y = f(\sum wx + b)$$

Where:
- **x** = Input values
- **w** = Weights
- **b** = Bias
- **f** = Activation Function

If the output is greater than a threshold, the perceptron classifies it as 1; otherwise 0. Binary classification means classifying data into two categories such as Yes/No, True/False, or Spam/Not Spam.

## Module 1 – Single-Layer Perceptron using NumPy

### Theory
NumPy is used for numerical operations and matrix calculations. In perceptron implementation, NumPy helps perform vector multiplication and weight updates efficiently. The perceptron learns by adjusting weights based on prediction errors.

### Common Syntax
```python
import numpy as np
np.dot(input, weights)
```

### Program

````python
import numpy as np

# Input dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Target output (AND Gate)
y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0
learning_rate = 0.1

# Activation Function
def step_function(x):
    return 1 if x >= 0 else 0

# Training
for epoch in range(10):
    for i in range(len(X)):
        linear_output = np.dot(X[i], weights) + bias
        prediction = step_function(linear_output)
        error = y[i] - prediction

        # Update weights and bias
        weights += learning_rate * error * X[i]
        bias += learning_rate * error

print("Weights:", weights)
print("Bias:", bias)

# Testing
for i in range(len(X)):
    output = step_function(np.dot(X[i], weights) + bias)
    print(X[i], "->", output)
````

## Module 2 – Single-Layer Perceptron using TensorFlow

### Theory
TensorFlow is an open-source deep learning framework used to build machine learning models. It provides built-in layers and optimizers for neural networks. A perceptron can be implemented using Dense layers in TensorFlow.

### Common Syntax
```python
import tensorflow as tf
model = tf.keras.Sequential()
```

### Program

````python
import tensorflow as tf
import numpy as np

# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

# Output data
y = np.array([[0], [0], [0], [1]], dtype=float)

# Create Model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, activation='sigmoid', input_shape=(2,))
])

# Compile Model
model.compile(
    optimizer='sgd',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train Model
model.fit(X, y, epochs=100, verbose=0)

# Predictions
predictions = model.predict(X)

print("Predictions:")
print(predictions)
````

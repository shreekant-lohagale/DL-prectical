## Theory
A Multi-Layer Perceptron (MLP) is a type of Artificial Neural Network (ANN) consisting of an **Input Layer**, **Hidden Layer(s)**, and **Output Layer**. MLP uses forward propagation and backpropagation for learning and is widely used for classification tasks.

The **MNIST dataset** contains 70,000 handwritten digit images from 0–9. Each image is 28×28 pixels.

Neuron output: **y = f(Σwx + b)**, where **x** is input, **w** is weights, **b** is bias, and **f** is the activation function.

Keras is a high-level deep learning API built on TensorFlow that simplifies neural network implementation.

## Module 1 – Loading and Preprocessing MNIST Dataset

**Theory:** MNIST contains grayscale handwritten digit images. Data preprocessing normalizes pixel values between 0 and 1. Flattening converts 28×28 images to one-dimensional vectors for MLP input.

**Common Syntax:**
```python
from tensorflow.keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

````python
from tensorflow.keras.datasets import mnist
import numpy as np

# Load Dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize Data
x_train = x_train / 255.0
x_test = x_test / 255.0

# Flatten Images
x_train = x_train.reshape(60000, 784)
x_test = x_test.reshape(10000, 784)

print("Training Data Shape:", x_train.shape)
print("Testing Data Shape:", x_test.shape)
````

## Module 2 – Building MLP Model using Keras

**Theory:** The Keras Sequential API creates neural networks layer by layer. Dense layers are fully connected. ReLU is used in hidden layers, and Softmax is used for multi-class classification.

**Common Syntax:**
```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create Model
model = Sequential()

# Hidden Layer
model.add(Dense(128, activation='relu', input_shape=(784,)))

# Output Layer
model.add(Dense(10, activation='softmax'))

# Model Summary
model.summary()
````

## Module 3 – Compiling and Training the Model

**Theory:** The model must be compiled before training. Adam optimizer improves learning efficiency, and sparse categorical crossentropy is used as the loss function for multi-class classification. Training adjusts weights to minimize loss.

**Common Syntax:** `model.compile()` and `model.fit()`.

````python
# Compile Model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train Model
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=32
)
````

## Module 4 – Evaluating the Model

**Theory:** Model evaluation checks performance on unseen test data. Accuracy indicates how correctly the model classifies handwritten digits.

**Common Syntax:** `model.evaluate()`.

````python
# Evaluate Model
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test Accuracy:", test_accuracy)
````

## Module 5 – Predicting Handwritten Digits

**Theory:** Prediction identifies the digit class from the trained model, which returns probabilities for each digit class from 0–9.

**Common Syntax:** `model.predict()`.

````python
import numpy as np

# Predict First Test Image
prediction = model.predict(x_test[:1])
print("Predicted Digit:", np.argmax(prediction))
print("Actual Digit:", y_test[0])
````

## Program — Step 1: Import Libraries

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

print("TensorFlow version:", tf.__version__)
```

## Step 2: Load Dataset

```python
vocab_size = 10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)
print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

## Step 3: Pad Sequences
Reviews have different lengths. Padding makes them equal. The document specifies a maximum length of 200.

```python
max_length = 200
X_train = pad_sequences(X_train, maxlen=max_length)
X_test = pad_sequences(X_test, maxlen=max_length)
print(X_train.shape)
```

## Step 4: Build RNN Model

```python
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=32, input_shape=(max_length,)))
model.add(SimpleRNN(32))
model.add(Dense(1, activation='sigmoid'))
```

## Step 5: Compile Model

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

## Step 6: Train Model
Training uses 5 epochs, batch size 64 and 20% validation split, as specified in the document.

```python
history = model.fit(X_train, y_train, epochs=5, batch_size=64, validation_split=0.2)
```

## Step 7: Evaluate Model

```python
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

## Step 8: Predict Sentiment

```python
prediction = model.predict(X_test[:1])
print(prediction)
if float(prediction[0][0]) > 0.5:
    print("Positive Review")
else:
    print("Negative Review")
```

## Improved Version Using LSTM
The document describes LSTM as an advanced RNN architecture that helps overcome the vanishing gradient problem.

```python
from tensorflow.keras.layers import LSTM

lstm_model = Sequential([
    Embedding(vocab_size, 32, input_shape=(max_length,)),
    LSTM(64),
    Dense(1, activation='sigmoid')
])
lstm_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
lstm_history = lstm_model.fit(X_train, y_train, epochs=5)
```
